```python
import numpy as np
from agg_operators import gamma_agg, seuil_agg, geometrique_agg, perplexite_agg

import matplotlib.pyplot as plt
from matplotlib.widgets import Slider

value = 1
low_value = 0.1
N= 100
list_score = np.ones(N) * value

res_gamma =[]
res_seuil = []
res_geometrique = []
for i in range(N):
    tmp = list_score.copy()
    tmp[i] = low_value
    res_gamma.append(gamma_agg(tmp))
    res_seuil.append(seuil_agg(tmp))
    res_geometrique.append(geometrique_agg(tmp))

fig, ax = plt.subplots(1,3, figsize=(25,5))
ax[0].plot(np.arange(N), res_gamma, marker='o', label='Gamma')
ax[0].set_xlabel('Index valeur faible')
ax[0].set_ylabel('Gamma Agg')
ax[0].set_title('Operateur gamma (temporel)')
ax[0].grid()

ax[1].plot(np.arange(N), res_seuil, marker='o', label='Seuil')
ax[1].set_xlabel('Index valeur faible')
ax[1].set_ylabel('Seuil Agg')
ax[1].set_title('Operateur seuil (temporel)')
ax[1].grid()

ax[2].plot(np.arange(N), res_geometrique, marker='o', label='Geometrique')
ax[2].set_xlabel('Index valeur faible')
ax[2].set_ylabel('Geometrique Agg')
ax[2].set_title('Operateur geometrique (temporel)')
ax[2].grid()
plt.show()


```
